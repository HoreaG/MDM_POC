```python
import pandas as pd
from mdm.conform import build_records
from mdm.deterministic import run_deterministic
from mdm.cluster import cluster
from mdm.scoring import score, summary

records = build_records("../data/sample")
pairs = run_deterministic(records)
clusters = cluster(records, pairs)

decisions = pd.read_csv("../data/sample/legacy_match_decisions.csv", dtype=str)
result = score(records, clusters, decisions)
summary(result)
```
